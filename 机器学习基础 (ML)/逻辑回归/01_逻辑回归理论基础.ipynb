# 逻辑回归基础

- 单一逻辑回归 - 即**神经元**(感知器/感知机)
- 逻辑回归的过程可以分为** 线性回归 ** -> ** Logit转化 ** -> ** 求最大似然概率下的样本可信度 ** -> ** 转化为Loss损失函数 ** -> ** 梯度下降递归求拟合的参数 **
- 在线性回归前，需要先了解特征矩阵与参数矩阵的行与列分别表示什么
    - 特征矩阵的列表示样本参数的维度，维度越高，表示的数据就越复杂。行表示样本数据的个数，如果样本数据过少，可能会导致欠定问题。但如果过于多，参数与样本数量过度不匹配又会导致欠拟合与过拟合的问题

## 1. 线性回归建模是所有逻辑回归的基础

- 在以逻辑回归为基础的神经网络世界中，线性回归均是以线性函数存在的分类函数
    - 数学公式:
    - $ Y = wx + b$ 其中w与b都是待拟合的参数
    - 后续的回归过程是不断修正拟合神经网络下各**神经元**的过程

## 2. 逻辑回归的建模是将线性回归推向梯度量化的关键

- 不同逻辑回归主要是根据分类种类不同和Logit转换函数不同进行划分的，这导致后续的流程依赖的函数有所不同
    - 主流逻辑回归的分别有:

### 2.1 二分逻辑回归的过程 - sigmoid下的Logit转化

- 模型在二分类的概念基础上搭建神经网络，样本硬性具有特征与真实标签两种属性，其中真实标签$y_i$具有二元性，即0或1
- 最常用的Logit变化: $ \rm{sigmoid}$
    - 一种将线性回归进行量化的数学公式假设推导，虽然是假设，但结论极其符合逻辑回归的拟合过程
    - 数学公式:
        - $ p = f(z) = \dfrac{1}{1 + e^{-z}}$
- 你只需要记住，sigmoid函数是将数值转化成概率的函数方式，这种激活函数有很多种，不同场景使用不同函数

### 2.2 二分逻辑回归的最大似然函数的建模反馈
- 理解为何要引入最大似然函数这一概念，对后续理解神经网络工作原理尤为重要
- 最大似然函数是用来反应当前参数下出现样本数据的真实概率。理论上，这个概率越高，代表越接近参数提供的拟合数据越接近真实数据，也就说明当前参数的置信度越高，这是衡量模型参数水平的最直观的标准
    -数学公式:
    - $ 符号不太好写，之后再补$
- 记住，最大似然是描述当前状态下参数对应的最大可能概率，这符合似然函数的标准定义

### 2.3 二分逻辑回归的损失函数的由来与作用
- 为什么会有损失函数，存在的意义是什么 --为了可以将极大似然函数转换成可以量化的值，并作用在后续的参数梯度下降中，有了损失函数，才有了梯度下降的方向，不至于每次调整参数都是摸着石头过河
- 损失函数的另一种形式的最大似然函数，可以将其理解为最大似然函数的另一种表现形式
- 最大似然函数的理论应用局限性:
    - 似然函数以累积的方式呈现，累积公式不利于后续计算梯度下降的求导运算
- 数学思维转换方式:
    - 在数学上，将累积转化为累加的方式是取对数，两者天然相等
    - 公式:
    - $ loss = ...$

### 2.4 二分逻辑回归的梯度下降函数
- 梯度下降则是一种根据损失函数确定拟合方向与改变数值的策略
- 直接上公式:
    - $ {w_New} = {w_Old} - w_i $
    - 这里的$w_i$是由损失函数对当前参数系数w求导得来的，公式的理论严谨性，由链式求导法则提供，$loss$对$w$的偏导，其实不难理解，$w_i$其实是参数对$p$的偏导值，因为$loss$是极大似然的对数形式，对$loss$求偏导即是极大似然的倒数，即每概率$p$下$w$的变化率，本质是参数变化的数值

## 3.梯度下降函数建模

```python
import numpy as np

def sigmoid(z):
    return 1/(1 + np.exp(-z))

def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)

logits = np.array([2.0, 1.0, 0.1])

print(sigmoid(logits))   # 相容（多类别）
print(softmax(logits))   # 排斥（单类别）
```
