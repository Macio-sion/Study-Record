## 规则分词 - 正则表达式分词

```python
import re
text1 = "fuck you bitch!"
text2 = """
Please contact support@example.com or call +1 (800) 555-0199.
My office number is 123-456-7890 ext. 1234.
You can also reach out to john.doe@company.co.uk.
The quick brown fox jumps over the lazy dog.
Don't forget about state-of-the-art technology.
"""
token_pattern = re.compile(
    r"""
    (?:                         # 非捕获组，用于匹配完整单元
        [A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}  # 1. 匹配邮箱
        |                       # 或
        (?:\+?\d{1,3}[-.\s]?)?  # 可选国际区号
        \(?\d{2,4}\)?           # 可选括号内的区号
        [-.\s]?                 # 可选分隔符
        \d{3,4}                 # 主号码前几位
        [-.\s]?                 # 可选分隔符
        \d{4}                   # 主号码后四位
        (?:[\s-]\d{3,6})?       # 可选分机号
        |                       # 或
        [A-Za-z]+(?:[-'][A-Za-z]+)*  # 3. 匹配英文单词（含连字符和缩写）
    )
    """,
    re.VERBOSE | re.IGNORECASE
)
tokens = token_pattern.findall(text2)
```

```python
tokens
```

## 统计分词 - 语义分词 - nltk

```python
import nltk
# nltk.download('punkt_tab')  # 需要先下载punkt_tab
from nltk.tokenize import word_tokenize
text = "I love NLP! It's great. Don't stop."
tokens = word_tokenize(text)
print(tokens)
```
